The `add` function currently takes two required arguments (`x`, `y`) and two optional arguments: `z` (defaults to 0) and `flag` (defaults to `False`).

```python
def add(x, y, z=0, flag=False):
    if flag:
        return x + y + z
    else:
        return x + y
```

- If `flag` is `True`, all three numbers (`x`, `y`, `z`) are added.
- If `flag` is `False` (or not provided), only `x` and `y` are added.

Extend the function to take another optional flag parameter and add only if it is true






````
def add(x, y, z=0, flag=False):

    if flag:
        return x + y + z

    else:
        return x + y

print(f"add(1, 2): {add(1, 2)}")
print(f"add(1, 2, 5): {add(1, 2, 5)}")
print(f"add(1, 2, 5, flag=True): {add(1, 2, 5, flag=True)}")
````

Now state the differences between a zip and enumerate iterator.

Both `enumerate()` and `zip()` are super helpful tools in Python for going through lists or other sequences.

### `enumerate()` - Your personal counter
- **What it does**: Imagine you have a list of items, like your favorite fruits. When you use `enumerate()`, it gives each fruit a number, starting from 0. So, it helps you keep track of *which* item you're looking at (its position) and *what* the item actually is.
- **When to use it**: Use `enumerate()` when you want to loop through a list and need to know both the item itself AND its position (index) in the list.
- **Example**: If you have `my_list = ['apple', 'banana', 'cherry']`, `enumerate(my_list)` would give you pairs like `(0, 'apple')`, `(1, 'banana')`, `(2, 'cherry')`. It's like saying "Apple is item number 0, Banana is item number 1," and so on.

### `zip()` - Pairing up lists
- **What it does**: Think of `zip()` like a zipper on a jacket. If you have two (or more) lists, `zip()` takes the first item from the first list, the first item from the second list, and zips them together into a pair. Then it does the same for the second items, and so on.
- **When to use it**: Use `zip()` when you have two or more lists that are related, and you want to go through them side-by-side, matching up their items.
- **Example**: If you have `names = ['Alice', 'Bob']` and `ages = [25, 30]`, `zip(names, ages)` would give you pairs like `('Alice', 25)` and `('Bob', 30)`. It connects corresponding items from different lists.

### Simple Differences:
- **`enumerate` is for *one* list**: It adds numbers to the items of a single list.
- **`zip` is for *multiple* lists**: It combines items from different lists, pairing them up like a team.
